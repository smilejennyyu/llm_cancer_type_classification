# Stratified Performance by Center

This notebook computes cancer type classification performance stratified by sequencing center for multiple models.

**Output:** For each model folder, saves `performance_stratified_by_center_{MODEL_FOLDER}.csv` to `genie_output/`.

```python
# ============================================================
# CONFIGURATION
# ============================================================
MODEL_FOLDERS = [
    "GDD-ENS",
    "gpt-5",
    "claude-opus-45",
    "deepseek-r1",
    "qwen3-vl-235b-a22b",
    "medgemma",
]

# Base paths
BASE_PATH = "/data1/morrisq/yuj13/llm_genomics"
GENIE_OUTPUT_PATH = f"{BASE_PATH}/genie_output"
```

```python
import numpy as np
import pandas as pd
import glob
import os
import unicodedata
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import matplotlib.ticker as mticker
from matplotlib import font_manager, rcParams

# Path to your font directory
font_dir = "/data1/morrisq/yuj13/share/fonts/liberation-fonts-ttf-2.1.5/"

# Register the font files with Matplotlib
font_manager.fontManager.addfont(font_dir + "LiberationSans-Regular.ttf")
font_manager.fontManager.addfont(font_dir + "LiberationSans-Bold.ttf")
font_manager.fontManager.addfont(font_dir + "LiberationSans-Italic.ttf")
font_manager.fontManager.addfont(font_dir + "LiberationSans-BoldItalic.ttf")

# Set Liberation Sans as the default font (Arial-compatible)
rcParams["font.family"] = "Liberation Sans"
```

```python
def normalize_text(name):
    """Normalize cancer type names by handling Unicode characters and punctuation."""
    if pd.isna(name):
        return "UNKNOWN"
    name = unicodedata.normalize('NFKD', str(name))
    name = name.replace('\u2011', '-')  # non-breaking hyphen
    name = name.replace('\u2010', '-')  # hyphen
    name = name.replace('\u2013', '-')  # en dash
    name = name.replace('\u2014', '-')  # em dash
    name = name.replace(',', '')  # remove commas
    return name.strip()


def detect_data_format(df):
    """Detect whether the data is from LLM (prediction1/prob1) or GDD-ENS (Pred1/Conf1) format."""
    if 'prediction1' in df.columns:
        return 'llm'
    elif 'Pred1' in df.columns and 'Conf1' in df.columns:
        return 'gdd'
    else:
        raise ValueError(f"Unknown data format. Columns: {df.columns.tolist()}")


def load_model_data(model_path, val_sample_ids=None):
    """
    Load and concatenate all CSV files from filtered_batch_msk and filtered_batch_non_msk.
    """
    msk_path = os.path.join(model_path, "filtered_batch_msk")
    non_msk_path = os.path.join(model_path, "filtered_batch_non_msk")
    
    dfs = []
    
    # Load non-MSK data
    non_msk_files = glob.glob(os.path.join(non_msk_path, "*.csv"))
    if non_msk_files:
        df_non_msk = pd.concat([pd.read_csv(f) for f in non_msk_files], ignore_index=True)
        dfs.append(df_non_msk)
        print(f"  Loaded {len(df_non_msk)} samples from non-MSK")
    
    # Load MSK data
    msk_files = glob.glob(os.path.join(msk_path, "*.csv"))
    if msk_files:
        df_msk = pd.concat([pd.read_csv(f) for f in msk_files], ignore_index=True)
        if val_sample_ids is not None:
            df_msk = df_msk[df_msk["SAMPLE_ID"].isin(val_sample_ids)]
        dfs.append(df_msk)
        print(f"  Loaded {len(df_msk)} samples from MSK")
    
    if not dfs:
        raise ValueError(f"No CSV files found in {model_path}")
    
    combined = pd.concat(dfs, ignore_index=True).drop_duplicates(subset="SAMPLE_ID")
    print(f"  Total unique samples: {len(combined)}")
    return combined
```

```python
def standardize_dataframe(df, data_format, mapping_df=None, clinical_df=None, full_mapping_df=None):
    """
    Standardize the dataframe to have consistent column names:
    - prediction1, prediction2: predicted cancer types
    - prob1, prob2: confidence scores
    - ground_truth: true cancer type
    
    For GDD format, uses CANCER_TYPE_DETAILED to correctly map predictions.
    """
    df = df.copy()
    
    if data_format == 'llm':
        # LLM format already has prediction1, prediction2, ground_truth
        df['prediction1'] = df['prediction1'].fillna('UNKNOWN').apply(normalize_text)
        df['prediction2'] = df['prediction2'].fillna('UNKNOWN').apply(normalize_text)
        
    elif data_format == 'gdd':
        # GDD format has Pred1, Conf1, Pred2, Conf2, Cancer_Type
        df = df.rename(columns={
            'Cancer_Type': 'ground_truth',
            'Conf1': 'prob1',
            'Conf2': 'prob2'
        })
        
        # For GDD-ENS, we need to use CANCER_TYPE_DETAILED for proper mapping
        if clinical_df is not None and full_mapping_df is not None:
            # Merge with clinical data to get CANCER_TYPE_DETAILED
            df = df.merge(
                clinical_df[['SAMPLE_ID', 'CANCER_TYPE_DETAILED']],
                on='SAMPLE_ID',
                how='left'
            )
            
            # Create simple mapping as fallback
            mapping = dict(zip(mapping_df['Cancer_Type'], mapping_df['CANCER_TYPE']))
            
            # Map Pred1 using CANCER_TYPE_DETAILED context
            df = df.merge(
                full_mapping_df[['CANCER_TYPE_DETAILED', 'Cancer_Type', 'CANCER_TYPE']],
                left_on=['CANCER_TYPE_DETAILED', 'Pred1'],
                right_on=['CANCER_TYPE_DETAILED', 'Cancer_Type'],
                how='left',
                suffixes=('', '_mapped1')
            )
            df = df.rename(columns={'CANCER_TYPE': 'prediction1'})
            df['prediction1'] = (
                df['prediction1']
                .fillna(df['Pred1'].map(mapping))
                .fillna('Unknown')
            )
            
            # Map Pred2 using CANCER_TYPE_DETAILED context
            df = df.merge(
                full_mapping_df[['CANCER_TYPE_DETAILED', 'Cancer_Type', 'CANCER_TYPE']],
                left_on=['CANCER_TYPE_DETAILED', 'Pred2'],
                right_on=['CANCER_TYPE_DETAILED', 'Cancer_Type'],
                how='left',
                suffixes=('', '_mapped2')
            )
            df = df.rename(columns={'CANCER_TYPE': 'prediction2'})
            df['prediction2'] = (
                df['prediction2']
                .fillna(df['Pred2'].map(mapping))
                .fillna('Unknown')
            )
        else:
            # Fallback to simple mapping if no clinical data
            mapping = dict(zip(mapping_df['Cancer_Type'], mapping_df['CANCER_TYPE']))
            df['prediction1'] = df['Pred1'].map(mapping).fillna(df['Pred1']).fillna('UNKNOWN')
            df['prediction2'] = df['Pred2'].map(mapping).fillna(df['Pred2']).fillna('UNKNOWN')
    
    return df
```

```python
def fix_cancer_type_names(df, ground_truth_set, verbose=True):
    """
    Fix common cancer type naming inconsistencies and mark unknown/invalid predictions.
    """
    df = df.copy()
    
    # Common replacements
    replacements = {
        'Skin Cancer Non-Melanoma': 'Skin Cancer, Non-Melanoma',
        'Non-Small-Cell Lung Cancer': 'Non-Small Cell Lung Cancer',
        'CNS Cancer (likely meningioma)': 'CNS Cancer',
        'Gastrointestinal Stromat Tumor': 'Gastrointestinal Stromal Tumor',
        'Gastrointestinal Stromatic Tumor': 'Gastrointestinal Stromal Tumor',
        'Gastrointestinal Stromenchymal Tumor': 'Gastrointestinal Stromal Tumor',
        'Malignant Mesothelioma': 'Mesothelioma',
        'Papillary Thyroid Cancer': 'Thyroid Cancer',
        'Gastrointestinal Stromen Tumor': 'Gastrointestinal Stromal Tumor',
        'Clear Cell Renal Cell Carcinoma': 'Renal Cell Carcinoma',
        'Papillary Cell Renal Cell Carcinoma': 'Renal Cell Carcinoma',
        'Urothelial Cancer (Bladder Cancer)': 'Bladder Cancer',
    }
    
    # Values that should be treated as UNKNOWN
    unknown_values = {'Indeterminate', 'Not Applicable', 'Unclassifiable', 'Undetermined', 'Unknown', 'UNKNOWN'}
    
    for col in ['prediction1', 'prediction2']:
        # Apply replacements
        df[col] = df[col].replace(replacements)
        
        # Mark unknown values
        df[col] = df[col].replace(unknown_values, 'UNKNOWN')
        
        # Mark predictions not in ground truth set as OTHERS
        invalid_preds = set(df[col].unique()) - ground_truth_set - {'UNKNOWN'}
        if invalid_preds and verbose:
            print(f"  Invalid predictions in {col} ({len(invalid_preds)}): {list(invalid_preds)[:5]}...")
        df[col] = df[col].replace(invalid_preds, 'OTHERS')
    
    return df
```

```python
def compute_stratified_performance(df, method_name):
    """
    Compute performance metrics stratified by CENTER.
    """
    required_cols = {'SAMPLE_ID', 'prediction1', 'prediction2', 'ground_truth', 'CENTER'}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(f"Missing required columns: {missing}")
    
    rows = []
    df_valid = df.dropna(subset=['CENTER', 'ground_truth']).copy()
    
    for center, group in df_valid.groupby('CENTER', dropna=False):
        y_true = group['ground_truth']
        y_pred1 = group['prediction1']
        y_pred2 = group['prediction2']
        
        n = len(group)
        if n == 0:
            continue
        
        acc = accuracy_score(y_true, y_pred1)
        top2_acc = ((y_pred1 == y_true) | (y_pred2 == y_true)).mean()
        w_precision = precision_score(y_true, y_pred1, average='weighted', zero_division=0)
        w_recall = recall_score(y_true, y_pred1, average='weighted', zero_division=0)
        w_f1 = f1_score(y_true, y_pred1, average='weighted', zero_division=0)
        
        rows.append({
            'CENTER': center,
            'Method': method_name,
            'N': n,
            'Accuracy': float(acc),
            'Top2_Accuracy': float(top2_acc),
            'Weighted_Precision': float(w_precision),
            'Weighted_Recall': float(w_recall),
            'Weighted_F1': float(w_f1),
            'CENTER_N': f"{center} (N={n})",
        })
    
    result = pd.DataFrame(rows)[
        ['CENTER', 'Method', 'N', 'Accuracy', 'Top2_Accuracy',
         'Weighted_Precision', 'Weighted_Recall', 'Weighted_F1', 'CENTER_N']
    ].sort_values(['CENTER', 'Method'], kind='stable')
    
    return result.reset_index(drop=True)
```

```python
# Load reference data (shared across all models)
print("Loading reference data...")
print("=" * 60)

# Load validation set for MSK filtering
val_set = pd.read_csv(f"{GENIE_OUTPUT_PATH}/o3-mini/output_nosite_extended_o3-mini.csv")
val_set["SAMPLE_ID"] = "GENIE-MSK-" + val_set["SAMPLE_ID"].astype(str)
val_sample_ids = set(val_set["SAMPLE_ID"])

# Load clinical data for CENTER mapping
data_clinical_sample = pd.read_csv(
    f"{BASE_PATH}/genie_data_v17.0/data_clinical_sample.txt",
    sep='\t', comment='#'
)

# Load assay info for SEQ_ASSAY_ID -> CENTER mapping
assay_df = pd.read_table(f"{BASE_PATH}/genie_data_v17.0/assay_information.txt")
seq_assay_to_center = dict(zip(assay_df['SEQ_ASSAY_ID'], assay_df['CENTER']))

# Load tumor type mapping (for GDD-ENS format)
# Keep both full mapping (for CANCER_TYPE_DETAILED context) and unique mapping (for simple lookup)
mapping_df = pd.read_table(f"{BASE_PATH}/GDD_ENS/data/tumor_type_final.txt")
mapping_df_unique = mapping_df[['CANCER_TYPE', 'Cancer_Type']].drop_duplicates()

print(f"Loaded {len(val_sample_ids)} validation sample IDs")
print(f"Loaded {len(data_clinical_sample)} clinical samples")
print(f"Loaded {len(seq_assay_to_center)} assay-to-center mappings")
print(f"Loaded {len(mapping_df)} tumor type mappings ({len(mapping_df_unique)} unique)")
print()
```

```python
def process_model(model_folder):
    """
    Process a single model folder and return performance results.
    """
    model_path = f"{GENIE_OUTPUT_PATH}/{model_folder}"
    
    print(f"\n{'='*60}")
    print(f"Processing: {model_folder}")
    print(f"{'='*60}")
    
    try:
        # Load data
        df = load_model_data(model_path, val_sample_ids)
        
        # Detect format and standardize
        data_format = detect_data_format(df)
        print(f"  Data format: {data_format}")
        
        # For GDD format, use full mapping with CANCER_TYPE_DETAILED
        if data_format == 'gdd':
            df = standardize_dataframe(
                df, data_format, 
                mapping_df=mapping_df_unique,
                clinical_df=data_clinical_sample,
                full_mapping_df=mapping_df
            )
        else:
            df = standardize_dataframe(df, data_format, mapping_df_unique)
        
        # Add CENTER from clinical data (if not already added during GDD processing)
        if 'SEQ_ASSAY_ID' not in df.columns:
            df = df.merge(
                data_clinical_sample[['SAMPLE_ID', 'SEQ_ASSAY_ID']],
                on='SAMPLE_ID',
                how='left'
            )
        df['CENTER'] = df['SEQ_ASSAY_ID'].map(seq_assay_to_center)
        
        # Get ground truth set and fix names
        ground_truth_set = set(df['ground_truth'].unique())
        df = fix_cancer_type_names(df, ground_truth_set)
        
        # Print summary
        unknown_count = (df['prediction1'] == 'UNKNOWN').sum()
        others_count = (df['prediction1'] == 'OTHERS').sum()
        print(f"  UNKNOWN: {unknown_count}, OTHERS: {others_count}, Centers: {df['CENTER'].nunique()}")
        
        # Compute performance
        result = compute_stratified_performance(df, model_folder)
        
        # Save results
        output_file = f"{GENIE_OUTPUT_PATH}/performance_stratified_by_center_{model_folder}.csv"
        result.sort_values(by=['Method', 'CENTER']).to_csv(output_file, index=False)
        print(f"  Saved to: {output_file}")
        
        return result
        
    except Exception as e:
        import traceback
        print(f"  ERROR: {e}")
        traceback.print_exc()
        return None
```

```python
# Process all models
all_results = {}

for model_folder in MODEL_FOLDERS:
    result = process_model(model_folder)
    if result is not None:
        all_results[model_folder] = result

print(f"\n\n{'='*60}")
print(f"SUMMARY: Processed {len(all_results)}/{len(MODEL_FOLDERS)} models successfully")
print(f"{'='*60}")
```

```python
# Combine all results into one dataframe
if all_results:
    combined_results = pd.concat(all_results.values(), ignore_index=True)
    combined_results = combined_results.sort_values(['CENTER', 'Method']).reset_index(drop=True)
    
    # Save combined results
    combined_output = f"{GENIE_OUTPUT_PATH}/performance_stratified_by_center_all_models.csv"
    combined_results.to_csv(combined_output, index=False)
    print(f"Combined results saved to: {combined_output}")
    
    # Display summary
    print(f"\nTotal rows: {len(combined_results)}")
    print(f"Models: {combined_results['Method'].nunique()}")
    print(f"Centers: {combined_results['CENTER'].nunique()}")
```

## Visualization: Performance with Data Availability and Cancer Type Distribution

```python
# Load features data for availability heatmap
import re

# Load features from genie_features directory
features_path_msk = f"{BASE_PATH}/genie_features/msk/filtered_batch"
features_path = f"{BASE_PATH}/genie_features/non_msk/filtered_batch"

# Load MSK features and filter to validation set
features_msk = pd.concat(
    (pd.read_csv(f) for f in glob.glob(os.path.join(features_path_msk, "*.csv"))),
    ignore_index=True
)
features_msk = features_msk[features_msk["SAMPLE_ID"].isin(val_sample_ids)]

# Load non-MSK features
features = pd.concat(
    (pd.read_csv(f) for f in glob.glob(os.path.join(features_path, "*.csv"))),
    ignore_index=True
)

# Combine
features = pd.concat([features, features_msk], ignore_index=True).drop_duplicates(subset="SAMPLE_ID")
print(f"Loaded {len(features)} feature samples")

# Load annotations to create category mapping
annotations = pd.read_csv(f"{BASE_PATH}/GDD_ENS/data/feature_annotations.csv")

def map_category(row):
    col = row['in_col']
    desc = row['out_col'].lower()
    
    # Mutations
    if 'mutation' in desc or col.endswith('_TRUNC') or col in ['LogSNV_Mb', 'LogINDEL_Mb']:
        if not col.lower().endswith('_hotspot'):
            return 'Mutations'
    
    # Hotspots
    if col.lower().endswith('_hotspot') or 'hotspot' in desc:
        return 'Hotspots'
    
    # Segmental CNAs (chromosome arm / region level)
    if col.startswith('Amp_') or col.startswith('Del_') or col == 'CN_Burden':
        return 'Segmental CNAs'

    # Gene CNAs (gene-level amplification / homozygous deletion)
    if col.endswith('_Amp') or col.endswith('_HomDel'):
        return 'Gene CNAs'
    
    # Fusions
    if col.lower().endswith('_fusion') or col.lower().endswith('_sv') or 'fusion' in desc:
        return 'Fusions'
    
    # Signatures
    if col.startswith('Sig_') or re.fullmatch(r'[ACGT]{4}', col):
        return 'Signatures'
    
    return 'Other'

annotations['Category'] = annotations.apply(map_category, axis=1)

# Create unique_features dataframe
unique_features = (
    annotations[['Category', 'in_col']]
    .drop_duplicates()
    .groupby('Category')['in_col']
    .apply(lambda items: sorted(set(items)))
    .reset_index()
)
unique_features.columns = ['Category', 'Features']

print(f"Feature categories: {unique_features['Category'].tolist()}")
```

```python
# Prepare data availability (pct_present_df) - boolean presence of feature categories per center
cat_map = dict(zip(unique_features['Category'], unique_features['Features']))

# Merge features with CENTER info
features_with_center = features.merge(
    data_clinical_sample[['SAMPLE_ID', 'SEQ_ASSAY_ID']],
    on='SAMPLE_ID',
    how='left'
)
features_with_center['CENTER'] = features_with_center['SEQ_ASSAY_ID'].map(seq_assay_to_center)

# Compute non-zero counts for each category
features_idx = features_with_center.set_index('SAMPLE_ID')
feat_counts = pd.DataFrame(index=features_idx.index)
for category, cols in cat_map.items():
    valid_cols = [c for c in cols if c in features_idx.columns]
    feat_counts[category] = (features_idx[valid_cols] != 0).sum(axis=1)

feat_counts['CENTER'] = features_idx['CENTER']

# Compute number of features in each category
n_feat = {cat: len(cols) for cat, cols in cat_map.items()}

# Compute mean non-zero count per institute
mean_counts = feat_counts.groupby('CENTER')[list(cat_map.keys())].mean()

# Convert to percentage of possible features
pct_present = mean_counts.div(pd.Series(n_feat)).multiply(100)

# Drop 'Other' and 'Signatures' if present, convert to boolean
cols_to_drop = [c for c in ['Other', 'Signatures'] if c in pct_present.columns]
pct_present_final = pct_present.drop(columns=cols_to_drop, errors='ignore').astype(bool)

# Reorder columns for display
desired_order = ['Mutations', 'Gene CNAs', 'Segmental CNAs', 'Hotspots', 'Fusions']
pct_present_final = pct_present_final[[c for c in desired_order if c in pct_present_final.columns]]

print("Data availability by center:")
display(pct_present_final)
```

```python
# Prepare cancer type distribution (topN_df)
# Use one of the processed model data to get ground truth distribution
# Load GDD-ENS data as reference for ground truth
gdd_path = f"{GENIE_OUTPUT_PATH}/GDD-ENS"
gdd_df = load_model_data(gdd_path, val_sample_ids)
gdd_df = gdd_df.rename(columns={'Cancer_Type': 'ground_truth'})
gdd_df = gdd_df.merge(
    data_clinical_sample[['SAMPLE_ID', 'SEQ_ASSAY_ID']],
    on='SAMPLE_ID',
    how='left'
)
gdd_df['CENTER'] = gdd_df['SEQ_ASSAY_ID'].map(seq_assay_to_center)

df_top_n = gdd_df[['SAMPLE_ID', 'CENTER', 'ground_truth']].dropna()
top_n = len(df_top_n.ground_truth.unique())

# Count samples by (CENTER, ground_truth)
counts = (df_top_n.groupby(['CENTER', 'ground_truth'])
            .size()
            .rename('n')
            .reset_index())

# Rank within each center
topN_df = (counts
         .assign(rank=lambda d: d.groupby('CENTER')['n'].rank(method='first',
                                                              ascending=False).astype(int))
         .query(f'rank <= {top_n}')
         .sort_values(['CENTER', 'n'], ascending=[True, False])
         .reset_index(drop=True))

# Add percent of that center's total
center_totals = df_top_n.groupby('CENTER').size().rename('center_total')
topN_df = (topN_df.merge(center_totals, on='CENTER')
              .assign(pct_of_center=lambda d: d['n'] / d['center_total'])
              .drop(columns='center_total'))

print(f"Cancer type distribution prepared with {len(topN_df)} entries")
display(topN_df.head(20))
```

```python
def plot_perf_with_availability_and_topN(
    df,
    pct_present_df=None,
    topN_df=None,
    topN=5,
    methods_to_compare=None, 
    methods_label=None,
    figsize=(22, 14),
    save_plots=False,
    output_prefix='',
    point_size=110,
    alpha_points=0.9,
    shade_by_N=True,
    shade_color='#8b5cf6',
    n_alpha_min=0.06,
    n_alpha_max=0.35,
    lane_pad=0.12,
    continuous_cmap='Greens',
    binary_colors=('#e5e7eb', '#10b981'),
    show_colorbar=True,
    availability_label='Data availability',
    show_availability_grid=True,
    grid_color='#9ca3af',
    grid_linewidth=0.8,
    grid_alpha=0.8,
    stack_metric='pct_of_center',
    topK_global=14,
    show_other=True,
    stack_width=0.8,
    legend_max_cols=4,
):
    """
    One figure per metric:
      0) N-per-center bar plot (always at the top)
      1) Vertical dot plot (methods vs. metric) with N-based lane shading
      2) Availability heatmap (boolean or continuous) sharing x-axis
      3) Stacked bar of topN cancer types per center sharing x-axis
    
    Accuracy and Top-2 Accuracy are displayed in 0-100% scale.
    Precision, Recall, and F1 are displayed in 0-1 scale.
    """
    # Metrics
    metrics = ['Accuracy', 'Top2_Accuracy', 'Weighted_Precision', 'Weighted_Recall', 'Weighted_F1']
    metric_labels = {
        'Accuracy': 'Accuracy',
        'Top2_Accuracy': 'Top-2 Accuracy',
        'Weighted_Precision': 'Weighted Precision',
        'Weighted_Recall': 'Weighted Recall',
        'Weighted_F1': 'Weighted F1'
    }
    
    # Metrics that should be displayed as percentages (0-100%)
    accuracy_metrics = {'Accuracy', 'Top2_Accuracy'}

    # Methods / colors / markers
    if methods_to_compare is None:
        methods_to_compare = df['Method'].unique().tolist()

    colors_methods = {
        'GDD-ENS': '#000000',
        'gpt-5': '#f97316',
        'claude-opus-45': '#3232a8',
        'claude-sonnet-37': '#3232a8',
        'deepseek-r1': '#ffb343',
        'deepseek': '#ffb343',
        'qwen3-vl-235b-a22b': '#8b5cf6',
        'qwen3': '#8b5cf6',
        'medgemma': '#10b981',
    }
    default_colors = ['#6b7280', '#374151', '#d97706', '#059669', '#dc2626', '#7c3aed', '#2563eb']
    for i, m in enumerate(methods_to_compare):
        if m not in colors_methods:
            colors_methods[m] = default_colors[i % len(default_colors)]

    markers_cycle = ['o', 's', '^', 'D', 'P', 'X', '*', 'v', '<', '>', 'h', 'H', 'd', 'p']
    method_markers = {m: markers_cycle[i % len(methods_to_compare)] for i, m in enumerate(methods_to_compare)}

    # Align per method & sort centers by first method's metric
    df_f = df[df['Method'].isin(methods_to_compare)].copy()
    centers = df_f['CENTER'].unique()
    method_dfs = {}
    for m in methods_to_compare:
        method_dfs[m] = (df_f[df_f['Method'] == m]
                         .set_index('CENTER').reindex(centers).reset_index())
    
    # Sort by first method's accuracy (descending)
    first = methods_to_compare[0] if 'gpt-5' not in methods_to_compare else 'gpt-5'
    if first not in method_dfs:
        first = methods_to_compare[0]
    order = method_dfs[first]['Accuracy'].argsort()[::-1]
    for m in methods_to_compare:
        method_dfs[m] = method_dfs[m].iloc[order].reset_index(drop=True)
    centers_sorted = method_dfs[first]['CENTER'].tolist()

    # Shading alpha from N
    N_vals = method_dfs[first]['N'].astype(float).values
    N_min, N_max = float(np.nanmin(N_vals)), float(np.nanmax(N_vals))
    denom = (N_max - N_min) if (N_max - N_min) != 0 else 1.0
    def n_to_alpha(n):
        t = (float(n) - N_min) / denom
        return n_alpha_min + t * (n_alpha_max - n_alpha_min)

    nC = len(centers_sorted)
    nM = len(methods_to_compare)
    x = np.arange(nC)
    jitter = 0.8 / max(nM, 3)

    # Availability prep
    avail_df = None
    avail_mat = None
    avail_rows = None
    is_continuous = False
    if pct_present_df is not None:
        avail_df = pct_present_df.copy()
        if 'CENTER' in avail_df.columns:
            avail_df = avail_df.set_index('CENTER')
        avail_df = avail_df.reindex(centers_sorted)
        avail_df = avail_df.replace({True: 1.0, False: 0.0})
        for c in avail_df.columns:
            avail_df[c] = pd.to_numeric(avail_df[c], errors='coerce')
        vals = avail_df.values.flatten()
        finite_vals = vals[np.isfinite(vals)]
        is_binary = np.all(np.isin(np.unique(np.round(finite_vals, 8)), [0.0, 1.0])) if finite_vals.size else True
        is_continuous = not is_binary
        if finite_vals.size and np.nanmax(finite_vals) > 1.0:
            norm_vals = avail_df.values / 100.0
        else:
            norm_vals = avail_df.values
        avail_mat = np.ma.masked_invalid(norm_vals).T
        avail_rows = list(avail_df.columns)

    # topN stacked bars prep
    stack_df = None
    cancer_order = []
    cancer_colors = {}
    if topN_df is not None:
        tn = topN_df.copy()
        required_cols = {'CENTER', 'ground_truth', stack_metric}
        missing = required_cols - set(tn.columns)
        if missing:
            raise ValueError(f"topN_df is missing required columns: {missing}")

        if stack_metric == 'pct_of_center' and tn[stack_metric].max() > 1.0:
            tn[stack_metric] = tn[stack_metric] / 100.0

        stack_df = (tn.pivot_table(index='CENTER', columns='ground_truth',
                                   values=stack_metric, aggfunc='sum', fill_value=0.0)
                      .reindex(centers_sorted).fillna(0.0))

        totals = stack_df.sum(axis=0).sort_values(ascending=False)
        if topK_global is not None and len(totals) > topK_global:
            keep = totals.head(topK_global).index.tolist()
            others = [c for c in stack_df.columns if c not in keep]
            stack_df = stack_df[keep].copy()
            if show_other and others:
                stack_df['Other'] = (tn[tn['ground_truth'].isin(others)]
                                     .groupby('CENTER')[stack_metric]
                                     .sum()
                                     .reindex(centers_sorted)
                                     .fillna(0.0).values)
                cancer_order = keep + ['Other']
            else:
                cancer_order = keep
        else:
            cancer_order = totals.index.tolist()

        # Color palette
        colors_to_use = plt.cm.tab20(np.linspace(0, 1, 20)).tolist()
        colors_to_use.pop(7)
        colors_to_use.pop(6)
        long_pal = (colors_to_use +
                    plt.cm.tab20b(np.linspace(0, 1, 20)).tolist() +
                    plt.cm.tab20c(np.linspace(0, 1, 20)).tolist())
        for i, ct in enumerate(cancer_order):
            cancer_colors[ct] = long_pal[i % len(long_pal)]
        if 'Other' in cancer_order:
            cancer_colors['Other'] = '#9ca3af'

    # Plotting loop
    for metric in metrics:
        # Determine if this metric should be displayed as percentage (0-100) or proportion (0-1)
        is_accuracy_metric = metric in accuracy_metrics
        
        both = (avail_df is not None) and (stack_df is not None)
        only_av = (avail_df is not None) and (stack_df is None)
        only_st = (avail_df is None) and (stack_df is not None)

        if both:
            fig = plt.figure(figsize=figsize, dpi=600)
            gs = fig.add_gridspec(4, 1, height_ratios=[1.2, 4, 1.25, 2.2], hspace=0.12)
            ax_n = fig.add_subplot(gs[0])
            ax = fig.add_subplot(gs[1], sharex=ax_n)
            ax_av = fig.add_subplot(gs[2], sharex=ax_n)
            ax_st = fig.add_subplot(gs[3], sharex=ax_n)
        elif only_av:
            fig = plt.figure(figsize=figsize, dpi=600)
            gs = fig.add_gridspec(3, 1, height_ratios=[1.4, 4, 1], hspace=0.12)
            ax_n = fig.add_subplot(gs[0])
            ax = fig.add_subplot(gs[1], sharex=ax_n)
            ax_av = fig.add_subplot(gs[2], sharex=ax_n)
            ax_st = None
        elif only_st:
            fig = plt.figure(figsize=figsize, dpi=600)
            gs = fig.add_gridspec(3, 1, height_ratios=[1.4, 4, 2.3], hspace=0.12)
            ax_n = fig.add_subplot(gs[0])
            ax = fig.add_subplot(gs[1], sharex=ax_n)
            ax_st = fig.add_subplot(gs[2], sharex=ax_n)
            ax_av = None
        else:
            fig = plt.figure(figsize=figsize, dpi=600)
            gs = fig.add_gridspec(2, 1, height_ratios=[1.4, 4], hspace=0.12)
            ax_n = fig.add_subplot(gs[0])
            ax = fig.add_subplot(gs[1], sharex=ax_n)
            ax_av = None
            ax_st = None

        title = f'{metric_labels[metric]} Across Models'
        fig.suptitle(title, fontsize=26, fontweight='bold', y=0.91)

        # Top panel: N-per-center bars
        bars = ax_n.bar(x, N_vals, width=0.8, edgecolor='white', linewidth=0.6, zorder=2, color='skyblue')
        ax_n.bar_label(
            bars,
            labels=[f'{v:,.0f}' if np.isfinite(v) else '' for v in N_vals],
            padding=3, fontsize=20, fontweight='bold'
        )
        ax_n.set_xlim(-0.5, nC - 0.5)
        ax_n.set_ylabel('Samples (N)', fontsize=26, fontweight='bold', labelpad=10)
        ax_n.tick_params(axis='x', which='major', bottom=False, labelbottom=False)
        ax_n.tick_params(axis='y', labelsize=24)
        ax_n.yaxis.set_major_formatter(mticker.ScalarFormatter(useMathText=True))
        ax_n.ticklabel_format(style='sci', axis='y', scilimits=(0, 0))
        ax_n.yaxis.offsetText.set_fontsize(18)
        ax_n.grid(True, axis='y', alpha=0.3)
        if np.isfinite(N_vals).any():
            ymax = np.nanmax(N_vals) * 1.3
            ax_n.set_ylim(0, ymax)

        # Dots (metrics) - fixed y-axis limits based on metric type
        if is_accuracy_metric:
            # Accuracy metrics: fixed 0-100% scale
            ax.set_ylim(0, 100)
        else:
            # Precision/Recall/F1: fixed 0-1 scale
            ax.set_ylim(0, 1.0)

        if shade_by_N:
            for j in range(nC):
                alpha = n_to_alpha(N_vals[j])
                left = j - 0.5 + lane_pad
                width = 1.0 - 2 * lane_pad
                ax.add_patch(plt.Rectangle((left, 0), width, ax.get_ylim()[1],
                                           facecolor=shade_color, alpha=alpha, lw=0, zorder=0))
        else:
            for j in range(nC):
                left = j - 0.5 + lane_pad
                width = 1.0 - 2 * lane_pad
                ax.add_patch(plt.Rectangle((left, 0), width, ax.get_ylim()[1],
                                           facecolor=shade_color, alpha=0.5, lw=0, zorder=0))

        for i, m in enumerate(methods_to_compare):
            xm = x + (i - (nM - 1) / 2) * jitter
            if is_accuracy_metric:
                # Accuracy metrics: multiply by 100 for percentage display
                ym = method_dfs[m][metric].values * 100.0
            else:
                # Precision/Recall/F1: keep as-is (0-1 scale)
                ym = method_dfs[m][metric].values
            mask = ~np.isnan(ym)
            label = methods_label[m] if methods_label else m
            ax.scatter(xm[mask], ym[mask], s=point_size, marker=method_markers[m],
                       c=colors_methods[m], edgecolor=colors_methods[m], linewidth=0.6,
                       alpha=alpha_points, label=label, zorder=2)

        # Set y-axis label based on metric type with extra padding
        if is_accuracy_metric:
            ax.set_ylabel(f'{metric_labels[metric]} (%)', fontsize=26, fontweight='bold', labelpad=10)
        else:
            ax.set_ylabel(f'{metric_labels[metric]}', fontsize=26, fontweight='bold', labelpad=10)
        
        ax.set_xticks(x)
        ax.set_xticklabels([])
        ax.tick_params(axis='x', which='major', bottom=False, labelbottom=False)
        ax.tick_params(axis='y', labelsize=24)
        ax.grid(True, alpha=0.3, axis='y')
        ax.legend(title='Models', loc='upper center', ncol=2, fontsize=16, title_fontsize=18)

        # Availability heatmap
        if ax_av is not None:
            _px = np.arange(nC + 1) - 0.5
            _py = np.arange(len(avail_rows) + 1) - 0.5

            if is_continuous:
                cmap = plt.get_cmap(continuous_cmap).copy()
                cmap.set_bad('#f3f4f6')
                im = ax_av.pcolormesh(_px, _py, avail_mat,
                                      vmin=0.0, vmax=1.0, cmap=cmap, zorder=0)
                if show_colorbar:
                    cbar = fig.colorbar(im, ax=ax_av, orientation='horizontal',
                                        pad=0.25, fraction=0.08)
                    cbar.set_label(f'{availability_label} (%)', fontsize=12)
                    cbar.set_ticks([0, .25, .5, .75, 1])
                    cbar.set_ticklabels(['0', '25', '50', '75', '100'])
            else:
                cmap = ListedColormap([binary_colors[0], binary_colors[1]])
                cmap.set_bad('#f3f4f6')
                ax_av.pcolormesh(_px, _py, avail_mat,
                                 vmin=0.0, vmax=1.0, cmap=cmap, zorder=0)

            ax_av.set_ylim(len(avail_rows) - 0.5, -0.5)
            ax_av.set_xlim(-0.5, nC - 0.5)
            ax_av.set_yticks(np.arange(len(avail_rows)))
            ax_av.set_yticklabels(avail_rows, fontsize=12)
            ax_av.set_xticks(x)
            ax_av.set_xticklabels([])
            ax_av.tick_params(axis='x', which='major', bottom=False, labelbottom=False)
            ax_av.tick_params(axis='y', labelsize=24)

            if show_availability_grid:
                ax_av.set_xticks(np.arange(-0.5, nC, 1), minor=True)
                ax_av.set_yticks(np.arange(-0.5, len(avail_rows), 1), minor=True)
                ax_av.grid(which='minor', color=grid_color, linewidth=grid_linewidth, alpha=grid_alpha)
                ax_av.tick_params(axis='x', which='major', bottom=False, labelbottom=False)
                ax_av.tick_params(axis='y', labelsize=24)
                ax_av.set_axisbelow(False)

            for s in ['top', 'right', 'left']:
                ax_av.spines[s].set_visible(False)
            ax_av.grid(False, which='major')

            if shade_by_N:
                ax.text(0.9, 0.02, "Top shading intensity = N (number of samples)", transform=ax.transAxes,
                        ha='right', va='bottom', fontsize=16, color='#4b5563')

        # Stacked bars of topN
        stack_handles = None
        if ax_st is not None:
            ax_st.set_xlim(-0.5, nC - 0.5)
            if stack_metric == 'pct_of_center':
                ax_st.set_ylim(0, 1.0)
                ax_st.set_ylabel('Cancer type\ndistribution (%)', fontsize=26, fontweight='bold', labelpad=10)
                ax_st.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'{int(v * 100)}'))
            else:
                ymax = (stack_df.sum(axis=1).max() if stack_df is not None else 1.0)
                ax_st.set_ylim(0, ymax * 1.05)
                ax_st.set_ylabel('Cancer type distribution', fontsize=26, fontweight='bold', labelpad=10)

            bottoms = np.zeros(nC)
            for ct in cancer_order:
                heights = stack_df[ct].values
                ax_st.bar(x, heights, bottom=bottoms, width=stack_width,
                          color=cancer_colors[ct], edgecolor='white', linewidth=0.6, label=ct)
                bottoms += heights

            ax_st.grid(True, axis='y', alpha=0.3)
            ax_st.set_xlabel('Center', fontsize=26, fontweight='bold')
            ax_st.set_xticks(x)
            ax_st.tick_params(axis='y', labelsize=24)
            ax_st.set_xticklabels(centers_sorted, rotation=90, ha='center', fontsize=24)

            stack_handles = [Patch(facecolor=cancer_colors[c], edgecolor='white', label=c) for c in cancer_order]
            fig.legend(
                handles=stack_handles, title='Cancer Types',
                loc='lower center', bbox_to_anchor=(0.5, -0.15),
                ncol=min(legend_max_cols, max(2, int(np.ceil(len(stack_handles) / 2)))),
                fontsize=24, title_fontsize=24, frameon=True, borderaxespad=0.3
            )

        top_pad = 0.10
        bot_pad = 0.18 if stack_handles is not None else 0.10
        fig.tight_layout(rect=[0.02, bot_pad, 0.98, 1 - top_pad])

        if save_plots:
            fn = f'{output_prefix}{metric.lower()}_perf_availability_top{topK_global}.pdf' \
                 if output_prefix else f'{metric.lower()}_perf_availability_top{topK_global}.pdf'
            plt.savefig(fn, dpi=600, bbox_inches='tight')
            print(f"Saved {metric} plot as {fn}")

        plt.show()
```
